# Import all libraries you need

```python
import numpy as np
import pandas as pd
#scipy.special for the sigmoid function expit()
import scipy.special
import matplotlib.pyplot
#ensure the plots are inside this notebook, not an external window
%matplotlib inline
import time
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler
import sklearn
import sklearn.model_selection
import sklearn.ensemble
import matplotlib.pyplot as plt
```

```python
def get_converted_data(data,xData,yData,trainPercShare,randomShuffle,scaling):
    data=data.sample(frac=1, random_state=1) #Shuffle data
    if randomShuffle==True:
        data=data.sample(frac=1) #Shuffle data random
    trainShare=round(len(data)*trainPercShare/100) #Calculate Number of Samples for trainings data
    #Define trainings and test data
    trainData=data[:trainShare]
    testData=data[trainShare:]
    xTrain=trainData.loc[:,xData].values
    yTrain=trainData[yData].values
    xTest=testData.loc[:,xData].values
    yTest=testData[yData].values
    #Scale data
    scaler=None
    if scaling==True:
        scaler=StandardScaler()
        try:
            scaler.fit(xTrain)
            xTrain=scaler.transform(xTrain)
        except:
            scaler.fit(xTest)
            xTest=scaler.transform(xTest)
        try:
            xTest=scaler.transform(xTest)
        except: 
            pass
    return(xTrain,yTrain,xTest,yTest,scaler)
```

# Import data

```python
###################################################### Specify data ##########################################################################

dataCL=pd.read_csv('Iris.csv')
xDataCL=['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
yDataCL='Species'

dataREG=pd.read_csv('hour.csv')
xDataREG=['season', 'yr', 'mnth', 'hr', 'weekday', 'workingday', 'weathersit', 'temp', 'hum', 'windspeed']
yDataREG='cnt'
```

# Prepare data

```python
trainPercShareCL=50
randomShuffleCL=True

[xTrainCL, yTrainCL, xTestCL, yTestCL, scalerCL]=get_converted_data(dataCL,xDataCL,yDataCL,trainPercShareCL,randomShuffleCL,True)

trainPercShareREG=50
randomShuffleREG=True

[xTrainREG, yTrainREG, xTestREG, yTestREG, scalerREG]=get_converted_data(dataREG,xDataREG,yDataREG,trainPercShareREG,randomShuffleREG,True)
```

# Classification

## CART - Classifier (Iris)

```python
clf=DecisionTreeClassifier(max_depth=3)
# Train model
clf.fit(xTrainCL,yTrainCL)
# Print accuracy
print('Train accuracy: '+str(round(clf.score(xTrainCL,yTrainCL)*100,1))+'%')
print('Test accuracy: '+str(round(clf.score(xTestCL,yTestCL)*100,1))+'%')
```

## Random Forest - Classifier (Iris)

```python
clf=RandomForestClassifier(max_depth=3)
# Train model
clf.fit(xTrainCL,yTrainCL)
# Print accuracy
print('Train accuracy: '+str(round(clf.score(xTrainCL,yTrainCL)*100,1))+'%')
print('Test accuracy: '+str(round(clf.score(xTestCL,yTestCL)*100,1))+'%')
```

## Neural Network - Classifier (Iris)

```python
clf=MLPClassifier(activation='logistic',solver='sgd')
# Train model
clf.fit(xTrainCL,yTrainCL)
# Print accuracy
print('Train accuracy: '+str(round(clf.score(xTrainCL,yTrainCL)*100,1))+'%')
print('Test accuracy: '+str(round(clf.score(xTestCL,yTestCL)*100,1))+'%')
```
